```python
asian_cars = df.loc[
    df['origin'] == 'Asia',
    ['vehicle_weight', 'model_year']
].head(7)

asian_cars
```

```python
X = np.array(asian_cars)
XT = X.T
X
```

```python
XT
```

```python
XTX = XT.dot(X)
XTX
```

```python
np.linalg.matrix_rank(XTX)
```

```python
XTXi = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
pre_w = X.dot(XTXi)
w = y.dot(pre_w)
```
